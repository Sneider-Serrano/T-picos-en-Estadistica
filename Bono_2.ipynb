# NumPy y Pandas en Ciencia de Datos: Uso, Ventajas y Desventajas para la Implementación de Algoritmos


## NumPy (*Numerical Python*)

### ¿Para qué se usa en ciencia de datos?

NumPy es la biblioteca fundamental para el cómputo numérico en Python. En ciencia de datos se utiliza principalmente para:

- Representar y manipular datos numéricos en forma de **arreglos multidimensionales (ndarrays)**.
- Realizar **operaciones matemáticas y estadísticas** de forma vectorizada (sin bucles explícitos).
- Servir como **base para otras bibliotecas** como Pandas, Scikit-learn, TensorFlow y SciPy, que internamente operan sobre arreglos NumPy.
- Ejecutar operaciones de **álgebra lineal** (multiplicación de matrices, descomposiciones) esenciales en algoritmos de Machine Learning.
- Generar **datos aleatorios** para simulaciones y pruebas de algoritmos.

###Ventajas

| Ventaja | Descripción |
|---|---|
| **Alto rendimiento** | Las operaciones vectorizadas se ejecutan en C/Fortran internamente, siendo hasta 100x más rápidas que los bucles de Python puro. |
| **Bajo consumo de memoria** | Un `ndarray` ocupa menos memoria que una lista de Python equivalente porque almacena datos de tipo homogéneo. |
| **Broadcasting** | Permite operar arreglos de distintas formas sin necesidad de replicar datos, ahorrando memoria y código. |
| **Amplio ecosistema** | Es el estándar de facto; prácticamente todas las bibliotecas de ML y ciencia de datos lo utilizan. |
| **Operaciones matriciales nativas** | Ideal para implementar desde cero regresión lineal, redes neuronales, PCA, etc. |

###Desventajas

| Desventaja | Descripción |
|---|---|
| **Solo datos homogéneos** | Todos los elementos de un arreglo deben ser del mismo tipo (e.g., todos `float64`). No es adecuado para mezclar texto y números. |
| **Sin etiquetas** | Los arreglos no tienen nombres de columnas ni índices descriptivos, lo que dificulta la interpretación de datos tabulares. |
| **Manejo de datos faltantes limitado** | No tiene soporte nativo para valores `NaN` en arreglos enteros; requiere conversiones o uso de `np.nan` solo con flotantes. |
| **Curva de aprendizaje en indexación avanzada** | El *fancy indexing* y el *broadcasting* pueden resultar confusos para usuarios nuevos. |

## Pandas

### ¿Para qué se usa en ciencia de datos?

Pandas está diseñado para la **manipulación y análisis de datos tabulares y de series temporales**. En ciencia de datos se usa para:

- **Cargar, limpiar y transformar** conjuntos de datos desde archivos CSV, Excel, SQL, JSON, entre otros.
- Realizar **exploración de datos (EDA)**: estadísticas descriptivas, conteos, distribuciones.
- **Filtrar, agrupar y agregar** datos con una sintaxis expresiva similar a SQL.
- Gestionar **datos faltantes**: detección, eliminación e imputación.
- **Preparar los datos** antes de alimentar algoritmos de Machine Learning (codificación, normalización, división train/test).
- Trabajar con **series temporales**: remuestreo, ventanas deslizantes, interpolación.

###Ventajas

| Ventaja | Descripción |
|---|---|
| **Datos heterogéneos** | Un `DataFrame` puede contener columnas de distintos tipos (int, float, str, datetime) en la misma estructura. |
| **Etiquetas descriptivas** | Índices y nombres de columnas facilitan el acceso, la interpretación y la depuración de datos. |
| **Manejo robusto de datos faltantes** | Métodos como `.fillna()`, `.dropna()` e `.interpolate()` hacen sencillo tratar valores ausentes. |
| **Integración con el ecosistema** | Compatible con Matplotlib, Seaborn, Scikit-learn, SQLAlchemy y muchas otras bibliotecas. |
| **Operaciones tipo SQL** | `groupby`, `merge`, `pivot_table` y `join` permiten análisis complejos con pocas líneas de código. |

###Desventajas

| Desventaja | Descripción |
|---|---|
| **Mayor consumo de memoria** | Los `DataFrames` tienen una sobrecarga considerable respecto a los arreglos NumPy puros, especialmente con columnas de tipo `object`. |
| **Rendimiento inferior en operaciones numéricas puras** | Para álgebra lineal o cómputo intensivo, NumPy (o CuPy en GPU) es significativamente más rápido. |
| **No escala a Big Data de forma nativa** | Para conjuntos de datos que no caben en RAM se requieren alternativas como Dask, Polars o PySpark. |
| **API extensa y a veces inconsistente** | La gran cantidad de métodos y las diferencias entre versiones pueden generar confusión (e.g., `.ix` deprecado, diferencias entre `.loc` e `.iloc`). |
| **Iteración fila a fila es lenta** | Usar bucles `for` sobre filas de un DataFrame es muy ineficiente; se debe preferir vectorización o `.apply()`. |
